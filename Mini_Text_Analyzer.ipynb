```python
import re
import json
import math
from collections import Counter
def tokenize(text):
    """Convert text into lowercase word tokens."""
    return re.findall(r"\b[a-zA-Z0-9]+\b", text.lower())
def analyze_text(text):
    """Calculate basic text statistics."""
    words = tokenize(text)
    frequencies = Counter(words)
    total_letters = sum(len(word) for word in words)
    return {
        "characters": len(text),
        "words": len(words),
        "unique_words": len(set(words)),
        "average_word_length": round(
            total_letters / len(words), 3
        ) if words else 0,
        "most_common_words": frequencies.most_common(5),
        "word_frequencies": dict(frequencies)
    }
def vectorize(words, vocabulary):
    """Represent text as a Bag-of-Words vector."""
    frequencies = Counter(words)
    return [frequencies[word] for word in vocabulary]
def cosine_similarity(vector_a, vector_b):
    """Calculate cosine similarity between two vectors."""
    dot_product = sum(
        a * b for a, b in zip(vector_a, vector_b)
    )
    magnitude_a = math.sqrt(sum(a * a for a in vector_a))
    magnitude_b = math.sqrt(sum(b * b for b in vector_b))
    if magnitude_a == 0 or magnitude_b == 0:
        return 0.0
    return round(
        dot_product / (magnitude_a * magnitude_b), 4
    )
def main():
    text_a = input("Enter the first text: ")
    text_b = input("Enter the second text: ")
    words_a = tokenize(text_a)
    words_b = tokenize(text_b)
    vocabulary = sorted(set(words_a) | set(words_b))
    vector_a = vectorize(words_a, vocabulary)
    vector_b = vectorize(words_b, vocabulary)
    stats_a = analyze_text(text_a)
    stats_b = analyze_text(text_b)
    similarity = cosine_similarity(vector_a, vector_b)
    result = {
        "text_a": {
            "statistics": stats_a,
            "vocabulary": vocabulary,
            "vector": vector_a
        },
        "text_b": {
            "statistics": stats_b,
            "vocabulary": vocabulary,
            "vector": vector_b
        },
        "cosine_similarity": similarity
    }
    print(json.dumps(result, indent=4))
if __name__ == "__main__":
    main()
```
