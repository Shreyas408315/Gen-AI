```python
from transformers import AutoModel, AutoConfig

model_name = "google-bert/bert-base-uncased"

# Load model with pretrained weights
pretrained_model = AutoModel.from_pretrained(model_name)

print("Model loaded using from_pretrained()")

print("Model Type:", pretrained_model.config.model_type)

print(
    "Total Parameters:",
    sum(p.numel() for p in pretrained_model.parameters())
)
```

```python
# Load configuration
config = AutoConfig.from_pretrained(model_name)

# Create model using configuration only
new_model = AutoModel.from_config(config)

print("Model created using from_config()")

print("Model Type:", new_model.config.model_type)

print(
    "Total Parameters:",
    sum(p.numel() for p in new_model.parameters())
)
```

```python
import torch

print("COMPARISON OF BOTH MODELS")
print("==========================")

print("\n1. Pretrained Model:")
print("Loaded using: AutoModel.from_pretrained()")
print("Contains pretrained weights from Google BERT.")
print("Has learned language representations.")

print("\n2. Newly Initialized Model:")
print("Loaded using: AutoModel.from_config()")
print("Contains randomly initialized weights.")
print("Has not learned pretrained language representations.")

# Compare a parameter from both models
pretrained_weight = next(
    pretrained_model.parameters()
).detach()

new_weight = next(
    new_model.parameters()
).detach()

print("\nAre the first parameter tensors equal?")
print(torch.equal(pretrained_weight, new_weight))
```
