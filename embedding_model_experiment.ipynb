```python
%%writefile minilm_embeddings.py

import torch
import numpy as np
import json
from pathlib import Path
from sentence_transformers import SentenceTransformer

# Model name
MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"

# Sample sentences
sentences = [
    "Artificial intelligence is transforming the world.",
    "Machine learning helps computers learn from data.",
    "Natural language processing enables computers to understand text.",
    "Deep learning is a subset of machine learning.",
    "Python is a popular programming language.",
    "The weather is beautiful today.",
    "I enjoy learning new technologies.",
    "Neural networks are inspired by the human brain.",
    "Data science combines statistics and programming.",
    "Google Colab provides free cloud computing resources."
]

# Select device
device = "cuda" if torch.cuda.is_available() else "cpu"

print("Loading MiniLM model...")

model = SentenceTransformer(MODEL_NAME, device=device)

# Generate embeddings
embeddings = model.encode(
    sentences,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

embeddings = np.asarray(embeddings, dtype=np.float32)

# Calculate dimensions
dimensions = embeddings.shape[1]

# Calculate memory requirements
parameter_memory = sum(
    p.numel() * p.element_size()
    for p in model.parameters()
)

embedding_memory = embeddings.nbytes

# Save embeddings
np.save("minilm_embeddings.npy", embeddings)

# Save model information
metadata = {
    "model": MODEL_NAME,
    "dimensions": int(dimensions),
    "sentences": len(sentences),
    "parameters": sum(p.numel() for p in model.parameters()),
    "model_memory_mb": parameter_memory / (1024 ** 2),
    "embedding_memory_kb": embedding_memory / 1024
}

with open("minilm_metadata.json", "w") as f:
    json.dump(metadata, f, indent=4)

print("\nMiniLM Embeddings Generated Successfully!")
print("Embedding Shape:", embeddings.shape)
print("Dimensions:", dimensions)
print("Model Memory (MB):", round(metadata["model_memory_mb"], 2))
print("Embedding Memory (KB):", round(metadata["embedding_memory_kb"], 2))
```

```python
%%writefile e5_embeddings.py

import torch
import numpy as np
import json
from sentence_transformers import SentenceTransformer

# Model name
MODEL_NAME = "intfloat/multilingual-e5-small"

# Sample sentences
sentences = [
    "Artificial intelligence is transforming the world.",
    "Machine learning helps computers learn from data.",
    "Natural language processing enables computers to understand text.",
    "Deep learning is a subset of machine learning.",
    "Python is a popular programming language.",
    "The weather is beautiful today.",
    "I enjoy learning new technologies.",
    "Neural networks are inspired by the human brain.",
    "Data science combines statistics and programming.",
    "Google Colab provides free cloud computing resources."
]

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Loading multilingual E5 model...")

model = SentenceTransformer(MODEL_NAME, device=device)

# Add E5 passage prefix
input_sentences = [
    "passage: " + sentence for sentence in sentences
]

# Generate embeddings
embeddings = model.encode(
    input_sentences,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

embeddings = np.asarray(embeddings, dtype=np.float32)

# Calculate dimensions
dimensions = embeddings.shape[1]

# Calculate memory
parameter_memory = sum(
    p.numel() * p.element_size()
    for p in model.parameters()
)

embedding_memory = embeddings.nbytes

# Save embeddings
np.save("e5_embeddings.npy", embeddings)

# Save metadata
metadata = {
    "model": MODEL_NAME,
    "dimensions": int(dimensions),
    "sentences": len(sentences),
    "parameters": sum(p.numel() for p in model.parameters()),
    "model_memory_mb": parameter_memory / (1024 ** 2),
    "embedding_memory_kb": embedding_memory / 1024
}

with open("e5_metadata.json", "w") as f:
    json.dump(metadata, f, indent=4)

print("\nE5 Embeddings Generated Successfully!")
print("Embedding Shape:", embeddings.shape)
print("Dimensions:", dimensions)
print("Model Memory (MB):", round(metadata["model_memory_mb"], 2))
print("Embedding Memory (KB):", round(metadata["embedding_memory_kb"], 2))
```

```python
%%writefile bge_m3_embeddings.py

import torch
import numpy as np
import json
from sentence_transformers import SentenceTransformer

# Model name
MODEL_NAME = "BAAI/bge-m3"

# Sample sentences
sentences = [
    "Artificial intelligence is transforming the world.",
    "Machine learning helps computers learn from data.",
    "Natural language processing enables computers to understand text.",
    "Deep learning is a subset of machine learning.",
    "Python is a popular programming language.",
    "The weather is beautiful today.",
    "I enjoy learning new technologies.",
    "Neural networks are inspired by the human brain.",
    "Data science combines statistics and programming.",
    "Google Colab provides free cloud computing resources."
]

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Loading BGE-M3 model...")

model = SentenceTransformer(MODEL_NAME, device=device)

# Generate embeddings
embeddings = model.encode(
    sentences,
    batch_size=4,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

embeddings = np.asarray(embeddings, dtype=np.float32)

# Calculate dimensions
dimensions = embeddings.shape[1]

# Calculate memory
parameter_memory = sum(
    p.numel() * p.element_size()
    for p in model.parameters()
)

embedding_memory = embeddings.nbytes

# Save embeddings
np.save("bge_m3_embeddings.npy", embeddings)

# Save metadata
metadata = {
    "model": MODEL_NAME,
    "dimensions": int(dimensions),
    "sentences": len(sentences),
    "parameters": sum(p.numel() for p in model.parameters()),
    "model_memory_mb": parameter_memory / (1024 ** 2),
    "embedding_memory_kb": embedding_memory / 1024
}

with open("bge_m3_metadata.json", "w") as f:
    json.dump(metadata, f, indent=4)

print("\nBGE-M3 Embeddings Generated Successfully!")
print("Embedding Shape:", embeddings.shape)
print("Dimensions:", dimensions)
print("Model Memory (MB):", round(metadata["model_memory_mb"], 2))
print("Embedding Memory (KB):", round(metadata["embedding_memory_kb"], 2))
```

```python
%%writefile comparison.py

import numpy as np
import pandas as pd
import json
import os

# Model files
models = {
    "MiniLM": ("minilm_embeddings.npy", "minilm_metadata.json"),
    "BGE-small": ("bge_embeddings.npy", "bge_metadata.json"),
    "Multilingual-E5": ("e5_embeddings.npy", "e5_metadata.json"),
    "BGE-M3": ("bge_m3_embeddings.npy", "bge_m3_metadata.json")
}

results = []

for model_name, (embedding_file, metadata_file) in models.items():

    # Check whether files exist
    if not os.path.exists(embedding_file):
        print(f"Skipping {model_name}: embedding file not found.")
        continue

    if not os.path.exists(metadata_file):
        print(f"Skipping {model_name}: metadata file not found.")
        continue

    # Load embeddings
    embeddings = np.load(embedding_file)

    # Load metadata
    with open(metadata_file, "r") as f:
        metadata = json.load(f)

    # Calculate actual embedding memory
    embedding_memory = embeddings.nbytes

    # Calculate memory per vector
    memory_per_vector = embedding_memory / len(embeddings)

    results.append({
        "Model": model_name,
        "Dimensions": embeddings.shape[1],
        "Number of Sentences": embeddings.shape[0],
        "Parameters": metadata["parameters"],
        "Model Memory (MB)": round(metadata["model_memory_mb"], 2),
        "Embedding Memory (KB)": round(embedding_memory / 1024, 2),
        "Memory Per Vector (KB)": round(memory_per_vector / 1024, 2),
        "Saved File Size (KB)": round(
            os.path.getsize(embedding_file) / 1024, 2
        )
    })

# Create comparison table
df = pd.DataFrame(results)

print("\n========== EMBEDDING MODEL COMPARISON ==========\n")

print(df.to_string(index=False))

# Save comparison to CSV
df.to_csv("comparison.csv", index=False)

# Save comparison to JSON
df.to_json("comparison.json", orient="records", indent=4)

print("\n==============================================")
print("Comparison completed successfully!")
print("Saved: comparison.csv")
print("Saved: comparison.json")
```
