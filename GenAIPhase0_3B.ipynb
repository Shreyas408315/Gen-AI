```python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])
```

```python
print("Shape of A:", A.shape)
print("Shape of B:", B.shape)
print("Transpose of A:", A.T)
print("Shape of AB:", (A@B).shape)
```

```python
Q = np.array([
    [1, 0],
    [0, 1]
])

K = np.array([
    [1, 0],
    [1, 1]
])
```

```python
print("Shape of Q:", Q.shape)
print("Shape of K:", K.shape)
print("Shape of KT:", (K.T).shape)
print("Matrix Multiplication of Q and K.T:", Q@K.T)
```
