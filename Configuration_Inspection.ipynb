```python
from transformers import AutoConfig

model_name = "google-bert/bert-base-uncased"

# Load pretrained configuration
config = AutoConfig.from_pretrained(model_name)

print("BERT Model Configuration")
print("-------------------------")

print("Model Type:", config.model_type)
print("Hidden Size:", config.hidden_size)
print("Number of Hidden Layers:", config.num_hidden_layers)
print("Number of Attention Heads:", config.num_attention_heads)
print("Vocabulary Size:", config.vocab_size)
```

```python
print("""
1. model_type:
   Identifies the model architecture.
   For BERT, the value is 'bert'.

2. hidden_size:
   Represents the dimension of hidden representations.
   BERT Base uses 768 dimensions.

3. num_hidden_layers:
   Represents the number of Transformer encoder layers.
   BERT Base contains 12 layers.

4. num_attention_heads:
   Represents the number of attention heads in each
   Transformer layer.
   BERT Base contains 12 attention heads.

5. vocab_size:
   Represents the number of tokens in the tokenizer's
   vocabulary.
   BERT Base has 30,522 vocabulary entries.
""")
```
