Q2

```python
A = np.array([3, 4])
B = np.array([1, 2])
print("Addition:", A + B)
print("Subtraction:", A - B)
print("Multipliction:", A*B)
print("Divison:", A/B)
dot = np.dot(A, B)
print("Dot Product:", dot)
print("||A||:", np.linalg.norm(A))
print("||B||:", np.linalg.norm(B))
norm_A = np.linalg.norm(A)
norm_B = np.linalg.norm(B)
similarity = dot / (norm_A * norm_B)
print("Cosine Similarity:", similarity)
```

Q5

```python
def cosine_similarity(a, b):
    dot = np.dot(a, b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    return dot / (norm_a * norm_b)

A = np.array([1, 2, 3])
B = np.array([4, 5, 6])
print("Cosine Similarity:", cosine_similarity(A, B))
```

```python
cat = np.array([0.9, 0.8, 0.7, 0.1])
dog = np.array([0.85, 0.75, 0.65, 0.2])
car = np.array([0.1, 0.2, 0.3, 0.9])
print("Cosine Similarity of cat and dog:", cosine_similarity(cat, dog))
print("Cosine Similarity of cat and car:", cosine_similarity(cat, car))
print("Cosine Similarity of dog and car:", cosine_similarity(dog, car))
```
