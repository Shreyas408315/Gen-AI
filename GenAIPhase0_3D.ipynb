```python
import numpy as np

A = np.array([1, 2, 3])
B = np.array([2, 4, 6])
C = np.array([6, 1, 2])
```

```python
print("Dot Product:", np.dot(A, B))
print("Euclidean Distance:", np.linalg.norm(A - B))
cosine_similarityAB = np.dot(A, B) / np.linalg.norm(A) / np.linalg.norm(B)
cosine_similarityAC = np.dot(A, C) / np.linalg.norm(A) / np.linalg.norm(C)
print("Cosine Similarity A and B:", cosine_similarityAB)
print("Cosine Similarity B and C:", cosine_similarityAC)
```
