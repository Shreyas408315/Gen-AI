```python
def token_stream(text):
    words = text.split()
    for word in words:
        yield word
text = "Hello how are you"
for token in token_stream(text):
    print(token)
```

Exercise 6

```python
import asyncio
import time
async def fetch_model():
    print("Starting model fetch...")
    await asyncio.sleep(2)
    print("Model fetch completed")
    return "Model Data"
async def fetch_database():
    print("Starting database fetch...")
    await asyncio.sleep(2)
    print("Database fetch completed")
    return "Database Data"
async def main():
    start_time = time.time()
    results = await asyncio.gather(fetch_model(), fetch_database())
    end_time = time.time()
    print(f"\nResults: {results}")
    print(f"Total time taken: {end_time - start_time:.2f} seconds")
await main()
```

Mini Challenge

```python
class LLMClient:
    def __init__(self, model: str = "demo-model", temperature: float = 0.7) -> None:
        self.model = model
        self.temperature = temperature
    def generate(self, prompt: str) -> str:
        if not prompt or not isinstance(prompt, str):
            raise ValueError("Prompt must be a non-empty string.")
        return f"Generating response using {self.model}..."
client = LLMClient(
    model="demo-model",
    temperature=0.7
)
result = client.generate("Explain RAG")
print(result)
```
